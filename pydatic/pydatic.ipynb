##Pydantic
 - a Python library used for data validation and parsing based on Python type hints.

#### In simple terms, it helps you define the shape of your data and automatically:
- Validate it (make sure it’s correct)
- Convert it (e.g., strings → integers)
- Provide clear errors if something is wrong
#### Why use it
- Ensures incoming data (like API requests) is correct
- Reduces manual validation code
- Works great with frameworks like FastAPI

```python
from dataclasses import dataclass
@dataclass
class Student():
    name: str
    age: int
    grade: str  
```

```python
mystudent=Student("Maha", 20, "A")
print(mystudent)
```

```python
mystudent=Student("Maha", 20, 20)
print(mystudent)
```

```python
# pydantic 
from pydantic import BaseModel
class Student1(BaseModel):
    name: str
    age: int
    grade: str
mystudent1=Student1(name="Maha", age=20, grade="A")
print(mystudent1)
```

```python
mystudent1=Student1(name="Maha", age=20, grade=20)
print(mystudent1)
```

## 2. Model  with Optional 
- add optional  Fields  using  python's Optional type

```python
from typing import Optional
class Employee(BaseModel):
    name: str
    age: int
    department: Optional[str] = None # Optional field with default value None
    salary: Optional[float] = None # Optional field with default value None
    is_active: Optional[bool] = True # Optional field with default value True

employee1 = Employee(name="John", age=30)
print(employee1)
```

```python
employee2=Employee(name="John", age=30, department="HR", salary=50000, is_active=False)
print(employee2)
```

```python
employee3=Employee(name="John", age=30, department="HR", salary=50000, is_active=0)
print(employee3)
```
