### Matplotlib 
- it is one of the most widely used plotting libraries in Python. It’s part of the Python ecosystem and is especially popular for creating static, animated, and interactive visualizations.

##### Matplotlib lets you:
- Plot graphs (line, bar, scatter, etc.)
- Customize axes, labels, and titles
- Create publication-quality figures
- Save plots as images (PNG, PDF, SVG)
- Extremely flexible and customizable
- Works well with libraries like NumPy and Pandas
- Standard tool for data science and research
